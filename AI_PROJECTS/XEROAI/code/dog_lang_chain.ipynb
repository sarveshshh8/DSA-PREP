````python
## QUESTION TO BE ASKED

#1. I want my labrador to have a shiny coat 
#2. how do i promote youthfullness in my german sheperd dog
````

````python
# Define the system message introducing the AI assistant's capabilities.
# system = "You are an expert Dog Assistant."
from langchain import PromptTemplate

# Define a placeholder for the user's input.
human = "{text}"

# Create a chat prompt consisting of the system and human messages.
prompt = ChatPromptTemplate.from_messages([("system", system), ("human", human)])

# Invoke the chat chain with the user's input.
chain = prompt | llm
text  = input("submit the query: ")
# response = chain.invoke({"text": text})
response = chain.invoke({"text": text,"tailblaze": tailblaze})

# response = chain.invoke({"text": "my labrador dog  is suffering from obesity?"})
````
