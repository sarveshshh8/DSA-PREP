```python
def insert_interval(intervals, newInterval):
    result = []  # To store the final merged intervals
    i = 0
    n = len(intervals)

    # Step 1: Add intervals that come before the new interval
    while i < n and intervals[i][1] < newInterval[0]:
        result.append(intervals[i])
        i += 1

    # Step 2: Merge intervals that overlap with the new interval
    while i < n and intervals[i][0] <= newInterval[1]:
        newInterval[0] = min(newInterval[0], intervals[i][0])
        newInterval[1] = max(newInterval[1], intervals[i][1])
        i += 1

    # Add the merged new interval
    result.append(newInterval)

    # Step 3: Add intervals that come after the new interval
    while i < n:
        result.append(intervals[i])
        i += 1

    return result
```

```python
# Recursive function to find the sum of 
# numbers from 0 to n
def findSum(n):
  
    # Base case 
    if n == 0:
        return 0
    
    # Recursive case 
    return n + findSum(n - 1)

n = 5
print(findSum(n))
```

```python
# list to represent the Airports
airports = ['A', 'B', 'C', 'D', 'E']

# dictionary to represent the lines between Airports
# and the time it will take
lines = {
    ('A', 'B') : 4,
    ('A', 'C') : 2,
    ('B', 'C') : 1,
    ('B', 'D') : 2,
    ('C', 'D') : 4,
    ('C', 'E') : 5,
    ('E', 'D') : 1,
}

# we choose Airport A as the starting airport 
# and Airport D as the destination
start = 'A'
end = 'D'
```

```python
# minimum function for dictionary,
# it will return the key who have the smallest value
def minimum(dict):
    min_key = list(dict.keys())[0]
    for i in list(dict.keys())[1:]:
        if dict[i] < dict[min_key]:
            min_key = i
    return(min_key)

def dijkstra(airports, lines, start, end):
    unexplored = {airport : float('inf') for airport in airports}

    # print(unexplored)
    unexplored[start] = 0
    while len(unexplored) != 0:
        explore = minimum(unexplored)
        # print("explore",explore)
        if explore == end:
            break
        else:
            for path in lines.items():
                if path[0][0] == explore:
                    if path[0][1] in unexplored.keys():
                        check_time = unexplored[path[0][0]] + path[1]
                        if check_time < unexplored[path[0][1]]:
                            unexplored[path[0][1]] = check_time
                elif path[0][1] == explore:
                    if path[0][0] in unexplored.keys():
                        check_time = unexplored[path[0][1]] + path[1]
                        if check_time < unexplored[path[0][0]]:
                            unexplored[path[0][0]] = check_time
            del unexplored[explore]

    return(unexplored[explore])
```

```python
print(dijkstra(airports, lines, start, end))
```

```python
def fun(input_list):
    a_max = max(input_list)
    A_min = min(input_list)
    length = len(input_list) // 2
    right = length
    left = 0

    while left <= right:
        o1 = input_list[left+1] - input_list[left] if left+1 < len(input_list) else 0
        o2 = input_list[left+2] - input_list[left] if left+2 < len(input_list) else 0
        
        if o1 <= o2:
            left += 1
            right -= 1
        elif o1 > o2:
            left += 1
            right += 1
            
    return a_max, A_min  # or any value depending on your logic
```
